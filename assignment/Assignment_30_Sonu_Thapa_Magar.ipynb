```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    OneHotEncoder,
    PowerTransformer,
    MinMaxScaler
)

numerical_pipeline=Pipeline([("imputer",SimpleImputer(strategy="median")),
                             ("power",PowerTransformer(method="yeo-johnson")),
                             ("scaler",MinMaxScaler())])
```

```python
categorical_pipeline=Pipeline([("encoder",OneHotEncoder(sparse_output=False,dtype = int))])
```

```python
preprocessor=ColumnTransformer([("num",numerical_pipeline,numerical_columns),
                                ("cat",categorical_pipeline,categorical_columns)])
```

```python
pipeline=Pipeline([("preprocessor",preprocessor)])
```
