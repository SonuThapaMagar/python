```python
from scipy.stats import skew,kurtosis,gaussian_kde
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Assignment 19 
# Skewness

a=[2,3,3,4,5,6,7,8,25]
b=[1,15,16,17,18,19,20,21,22]
c=[10,12,15,15,15,16,18,20]


s_a=skew(a)
s_b=skew(b)
s_c=skew(c)


print(s_a)
print(s_b)
print(s_c)

g_a=gaussian_kde(a)
g_b=gaussian_kde(b)
g_c=gaussian_kde(c)

plt.figure(figsize=(9,6))

xa_range=np.linspace(min(a)-2,max(a)+2,100)
ya_range=g_a(xa_range)
plt.plot(xa_range, ya_range, color="red", label="Dataset A")

xb_range=np.linspace(min(b)-2,max(b)+2,100)
yb_range=g_b(xb_range)
plt.plot(xb_range, yb_range, color="green", label="Dataset B")

xc_range=np.linspace(min(c)-2,max(c)+2,100)
yc_range=g_c(xc_range)
plt.plot(xc_range, yc_range, color="blue", label="Dataset C")

plt.title("Gaussian KDE of Datasets")
plt.xlabel("Data Values")
plt.ylabel("Density")

plt.legend()
plt.grid(True)

plt.show()
```

```python
#Kurtosis
a=[4.2,5.1,4.8,5.5,4.9,5.0,5.3,4.7,5.2,4.6,5.4,4.5,5.1,4.9,5.0,5.2,4.8,5.3,4.7,5.1]

b=[5.0,5.1,4.9,5.0,5.05,4.95,5.0,5.02,4.98,5.0,5.01,4.99,5.0,2.0,8.0,1.5,8.5,5.0,5.0,5.0]

c=[3.0,3.5,4.0,4.5,5.0,5.5,6.0,6.5,7.0,3.2,3.8,4.2,4.8,5.2,5.8,6.2,6.8,7.2]


k_a=kurtosis(a)
k_b=kurtosis(b)
k_c=kurtosis(c)

g_a=gaussian_kde(a)
g_b=gaussian_kde(b)
g_c=gaussian_kde(c)

plt.figure(figsize=(9,6))

xa_range=np.linspace(min(a)-2,max(a)+2,100)
ya_range=g_a(xa_range)
plt.plot(xa_range, ya_range, color="red", label="Dataset A")

xb_range=np.linspace(min(b)-2,max(b)+2,100)
yb_range=g_b(xb_range)
plt.plot(xb_range, yb_range, color="green", label="Dataset B")

xc_range=np.linspace(min(c)-2,max(c)+2,100)
yc_range=g_c(xc_range)
plt.plot(xc_range, yc_range, color="blue", label="Dataset C")

plt.title("KDE for Kurtosis Comparison")
plt.xlabel("Data Values")
plt.ylabel("Density")

plt.legend()
plt.grid(True)

plt.show()
```

```python
#Assignment 20
""" Find the Z-scores for your marks on MCQ1, MCQ2, and MCQ3, treating 
the corresponding results as raw data."""
mcq_1=9
mcq_2=14
mcq_3=7

marks=np.array([9,14,7])
#Mean
mean=np.mean(marks)
print(f"Mean: {mean}")

#Standard Deviation
std=np.std(marks)
print(f"Standard Deviation: {std:.2f}")

#Z scores
z_score_mcq_1=(9-mean)/std
print(f"Z score of MCQ 1: {z_score_mcq_1:.2f}")

z_score_mcq_2=(14-mean)/std
print(f"Z score of MCQ 2: {z_score_mcq_2:.2f}")

z_score_mcq_3=(7-mean)/std
print(f"Z score of MCQ 3: {z_score_mcq_3:.2f}")

#Z scores
z_scores = [z_score_mcq_1, z_score_mcq_2, z_score_mcq_3]

mcqs = ["MCQ 1", "MCQ 2", "MCQ 3"]

plt.figure(figsize=(10,6))
plt.plot(mcqs, z_scores, marker="o", color="red", linewidth=2)
plt.title("Z-Scores of MCQ Marks")
plt.xlabel("MCQ")
plt.ylabel("Z-Score")

plt.grid(True)

plt.show()
```

```python
#Find the 40%, 60%, 80% percentile from MCQ1, MCQ2, MCQ3 result respectively.
marks=np.array([9,14,7])

p_40=np.percentile(marks,40)
p_60=np.percentile(marks,60)
p_80=np.percentile(marks,80)

print(f"40th Percentile : {p_40:.2f}")
print(f"60th Percentile : {p_60:.2f}")
print(f"80th Percentile : {p_80:.2f}")

percentiles=["40%","60%","80%"]
values=[p_40,p_60,p_80]

plt.figure(figsize=(8,5))
plt.bar(percentiles, values)

plt.title("Percentiles of MCQ Marks")
plt.xlabel("Percentile")
plt.ylabel("Marks")

plt.grid(axis="y")
plt.show()
```

```python
#Find the Covarience and correlation of the given data points:
a = np.array([2, 4, 5, 6, 7, 8, 9, 10])
b = np.array([55, 60, 65, 70, 75, 80, 85, 90] )

# Covariation
covariation_first=np.cov(a,b)
print(f"Covariance: {covariation_first}")

#Correlation
correlation_first=np.corrcoef(a,b)
print(f"Correlation:{correlation_first}")

plt.figure(figsize=(10,6))
plt.scatter(a,b,color="Red",label='Data Points')
plt.title("Scatter Plot of Data A and B")

plt.xlabel("Data A")
plt.ylabel("Data B")

plt.grid(True)
plt.legend()

plt.show()
```
