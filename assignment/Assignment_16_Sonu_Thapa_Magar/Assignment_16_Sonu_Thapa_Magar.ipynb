```python
#File Handling
"""
DictReader, DictWriter
Explore (read, write, append) only method.
"""
#Writing data in CSV File Handling

#Importing csv module
import csv

#Column headers
fieldnames=["name","age","marks"]

#List of dictionaries
students=[
    {"name":"Sonu","age":23,"marks":40},
    {"name":"Monu","age":32,"marks":45},
    {"name":"Shownu","age":43,"marks":67}
]

#Writing data
with open("students.csv",mode="w",) as f:
    writer=csv.DictWriter(f,fieldnames)
    #writing the header
    print(f.tell())
    writer.writeheader()
    print(f.tell())

    #writing the data
    writer.writerows(students)
    print(f.tell())
```

```python
#Reading data in CSV File Handling

#Importing csv module
import csv

#Reading data
with open("students.csv",mode="r",) as f:
    print(f.tell())
    reader=csv.DictReader(f)

    for row in reader:
        print(f"Name: {row['name']} | Age: {row['age']} | Marks: {row['marks']}") 
    print(f.tell())       
```

```python
#Appending Data in CSV
import csv

#Column headers
fieldnames=["name","age","marks"]

#List of dictionaries
new_students=[
    {"name":"Alex","age":18,"marks":49},
    {"name":"Alice","age":21,"marks":76},
    {"name":"Sean","age":24,"marks":65}
]

#Appending data
with open("students.csv",mode="a",) as f:
    print(f.tell())
    writer=csv.DictWriter(f,fieldnames)
    #writing the header
    writer.writeheader()

    #writing the data
    writer.writerows(new_students)
    print(f.tell())
```

```python
#Assignment 2
"""
Make text.txt file and write number from 1 to 50.
And use read mode to read that file and print sum.
"""
#write mode
with open("text.txt", "w") as file:
    for i in range(1, 51):
        file.write(str(i) + "\n")
```

```python
# Step 2: Read the file and calculate the sum
total = 0

with open("text.txt", "r") as file:
    for line in file:
        total += int(line.strip())

# Step 3: Print the sum
print("Sum =", total)
```

```python
#Assignment 3
"""
Take user input of (name, roll no, marks (only one)) of the student and save it into csv file.
Use read mode to find the highest mark student and print highest mark student info.
"""
import csv

#write
with open("new_students.csv",mode="w",newline="") as file:
    writer=csv.writer(file)
    writer.writerow(['Name','Roll No','Mark'])

    for i in range(5):
        name=input("Enter the name: ")
        roll_no=int(input("Enter the roll no: "))
        mark=int(input("Enter the mark: "))

        writer.writerow([name,roll_no,mark])
```

```python
#read
highest_student=None
highest_marks = -1

with open("new_students.csv",mode="r") as file:
    reader=csv.reader(file)

    next(reader)
    for row in reader:
        if len(row)<3:
            continue

        if int(row[2]) > highest_marks:
            highest_marks = int(row[2])
            highest_student = row

print("\nHighest Mark Student")
print("Name :", highest_student[0])
print("Roll :", highest_student[1])
print("Marks:", highest_student[2])
```
