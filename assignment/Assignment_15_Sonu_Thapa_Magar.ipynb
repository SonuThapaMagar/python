```python
"""
Question 1: __init__, __str__, and __setitem__
Create a Student class with attributes name, roll_no, and marks. Implement __init__ to initialize
values, __str__ to print the object in a readable format, and __setitem__ to update attributes using
dictionary-style syntax such as student['marks']=90. Demonstrate each method.
"""
class Student:
    def __init__(self,name,roll_no,marks):
        self.name=name
        self.roll_no=roll_no
        self.marks=marks

    def __str__(self):
        return (
            f"Name:{self.name}\n"
            f"Roll No:{self.roll_no}\n"
            f"Makrs: {self.marks}"
        )

    def __setitem__(self, key, value):
        if key == "name":
            self.name = value
        elif key == "roll_no":
            self.roll_no = value
        elif key == "marks":
            self.marks = value
        else:
            print("Invalid attribute")
```

```python
student=Student("Sonu",123,56)
student["marks"] = 90
student["name"] = "Ram"

print("\nAfter Updating:\n")
print(student)
```

```python
"""
Question 2: __add__
Create a Vector class representing a 2D vector (x, y). Implement __add__ so two vectors can be
added using the + operator. Print the resulting vector.
"""
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y

    def __add__(self, other):
        result1=self.x+other.x
        result2=self.y+other.y
        v3=Vector(result1,result2)
        return v3

    def __str__(self):
        return (
            f"Vector: {self.x}, {self.y}"
        )
```

```python
v1 = Vector(2, 3)
v2 = Vector(4, 5)
v3=v1+v2
print(v3)
```

```python
"""
Question 3: __eq__
Implement __eq__ in the Student class so two students are equal if they have the same roll
number. Test the comparison with different objects.
"""
class Student:
    def __init__(self,roll_no):
        self.roll_no=roll_no

    def __eq__(self, other):
        return self.roll_no==other.roll_no

    def __str__(self):
        return f"{self.roll_no}"
```

```python
s1=Student(101)
s2=Student(100)
s3=Student(101)
```

```python
print(s1==s2)
print(s2==s3)
print(s1==s3)
```

```python
"""
Question 4: __len__ and __getitem__
Create a Playlist class that stores a list of songs. Implement __len__ to return the total number of
songs and __getitem__ to access songs using indexing.
"""
class Playlist:
    def __init__(self,songs):
        self.songs=songs

    def __len__(self):
        return len(self.songs)

    def __getitem__(self, key):
        return self.songs[key]
```

```python
playlist = Playlist(["Believer", "Shape of You", "Perfect","Human","Snowman"])
```

```python
print("Total Songs: ",len(playlist))
print("First Song: ",playlist[0])
print("Last Song: ",playlist[-1])
```

```python
"""
Question 5: __call__
Create a Multiplier class that accepts a number in the constructor. Implement __call__ so the object
behaves like a function and returns the multiplication result
"""
class Multiplier:
    def __init__(self,number):
        self.number=number
        
    def __call__(self, value):
        return self.number*value
```

```python
multiplier=Multiplier(5)
print(multiplier(10))
print(multiplier(4))
print(multiplier(20))
```
