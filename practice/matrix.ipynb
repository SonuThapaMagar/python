```python
import numpy as np

#student data
X = np.array([
    [2, 70],
    [5, 80],
    [8, 90]
])

# Model weights
W = np.array([
    [3],
    [0.5]
])

#Bias
b=2

#Prediction
y_pred=X @ W+b

print(y_pred)
```

```python
print("X shape:", X.shape)
print("W shape:", W.shape)
print("Prediction shape:", y_pred.shape)
```
