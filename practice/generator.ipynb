```python
def even():
    for i in range(10):
        if i%2==0:
            yield i

for i in even():
    print(i)
```

```python
def normal():
    print("Before return")
    return 1
    print("After return")


def generator():
    print("Before yield")
    yield 1
    print("After yield")


print(normal())

g = generator()

print(next(g))
```
