YOLO

```python
from ultralytics import YOLO
import os
```

```python
model=YOLO("yolo11n.pt")
image_paths = [
    "../../datasets/PetImages/Cat/0.jpg",
    "../../datasets/PetImages/Cat/1.jpg",
    "../../datasets/PetImages/Dog/0.jpg",
    "../../datasets/PetImages/Dog/1.jpg",
    "../../datasets/PetImages/Dog/2.jpg"
]
```

```python
os.makedirs("output", exist_ok=True)
results=model(image_paths)
```

```python
print(model.names)
```

```python
for index,result in enumerate(results):
    save_path = f"output/result_{index}.jpg"
    result.save(filename=save_path)
    
    for box in result.boxes:
        x1, y1, x2, y2 = box.xyxy[0].tolist()
        confidence = box.conf[0].item()
        class_id = int(box.cls[0].item())
        class_name = model.names[class_id]
        
        print(f"Detected {class_name} with {confidence:.2f} confidence at [{x1:.1f}, {y1:.1f}, {x2:.1f}, {y2:.1f}]")
```
