```python
from ultralytics import YOLO
import torch
```

```python
dataset_path = "../../image_dataset/config.yaml"
```

Device setup

```python
device=0 if torch.cuda.is_available() else "cpu"

print("GPU available",torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("Using CPU")
```

Loading the base model

```python
model = YOLO("yolo26n.pt")
```

Training the model

```python
model.train(
    data=dataset_path,
    epochs=10,
    imgsz=224,
    batch=16,
    device=device,
    project="yolo_runs",
    name="train_yolo" 
)
```

```python
import os

project_root = r"C:\Users\USER\Desktop\python"

for root, dirs, files in os.walk(project_root):
    if "best.pt" in files:
        print(os.path.abspath(os.path.join(root, "best.pt")))
```

Loading Best Model for validation

```python
best_model = YOLO(
    r"C:\Users\USER\Desktop\python\runs\detect\yolo_runs\train_yolo-8\weights\best.pt"
)
```

```python
metrics = best_model.val(
    data=dataset_path,
    imgsz=224,
    batch=16,
    device=device
)
```

```python
print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
```
