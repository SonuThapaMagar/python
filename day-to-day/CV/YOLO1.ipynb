```python
from ultralytics import YOLO
import torch
import os
```

```python
BASE_DIR = os.path.dirname(os.path.abspath(__file__)) 
dataset_path = os.path.join(BASE_DIR, "..", "datasets", "image_dataset")
yaml_path = os.path.join(dataset_path, "config.yaml")
```

Device setup

```python
device=0 if torch.cuda.is_available() else "cpu"
print("GPU available",torch.cuda.is_available())
if torch.cuda.is_available():
    print(torch.cuda.get_device_name(0))
```

```python
model=YOLO("yolo11n.pt")
```

```python
model.train(data=yaml_path,epochs=10,imgsz=640,batch=16,device=device,project="",name="train_yolo")
```

```python
best_model=YOLO()
metrics=best_model.val()
```

```python
print("Precision: ",metrics.box.mp)
print("Recall: ",metrics.box.mr)
print("mAp50: ",metrics.box.map50) #mean average precision
print("mAp50-90: ",metrics.box.map)
```
