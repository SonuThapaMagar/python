Transfer Learning

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, random_split
from torchvision import transforms, datasets
import torchvision.models as models
```

```python
device=torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device",device)
```

```python
data="../dataset'"
```

```python
transformation=transforms.Compose([
    transforms.Resize((224,224)),
    transforms.RandomHorizontalFlip(p=0),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.5,0.5,0.5], #for channel 0.5-R,g,b
        std=[0.25,0.25,0.25]
    )
])
```

```python
dataset=datasets.ImageFolder(root=data,transform=transformation)
```

```python
print(dataset.class_to_idx)
print(len(dataset))
```

```python
train_set=int(0.8*len(dataset))
test_set=len(dataset)-train_set

print(train_set)
print(test_set)
```

```python
train_dataset,test_dataset=random_split(dataset,[train_set,test_set])
```

```python
train_loader=DataLoader(train_dataset,batch_size=16,shuffle=True)
```

```python
test_loader=DataLoader(train_dataset,batch_size=16,shuffle=False)
```

```python
model=models.resnet18(weights=models.ResNet18_Weights.DEFAULT)

for param in model.parameters():
    param.requires_grad=False
    
num_features=model.fc.in_features
model.fc=nn.Linear(num_features,2)
model=model.to(device)
```

```python
loss_fxn=nn.CrossEntropyLoss()
optimizer=torch.optim.Adam(model.parameters(),lr=0.001)
```
