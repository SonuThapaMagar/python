```python
#File Handling in python
#Write Mode with buffering =-1
file=open("test.txt",mode="w",buffering=-1)
file.write("Hello\n")
file.close()
```

```python
#Write Mode with buffering = 1 works in text mode . if there is \n it writes before upto \n
file=open("test.txt",mode="w",buffering=1)
file.write("Learning Python\n") #after breaking line only it starts writing when 1 is used
file.write("Its fun")
# file.close()

file.close()
```

```python
with open("abc.txt",mode="w",buffering=-1) as f: #performs open and close without needing of close()
    f.write("Hello World\n")
    f.write("I'm learning Python\n")
```

```python
with open("abc.txt",mode="a") as f:
    f.write("Learning file handling\n")
```

```python
with open("abc.txt",mode="w",buffering=-1) as f: #performs open and close without needing of close()
    content=["Hi\n","Hello\n","How are you?\n"]

    f.writelines(content) # multiple line writing
```

```python
#Read file
with open("abc.txt",mode="r") as f:
    # content=f.read()
    # content=f.readline()
    content=f.readlines() #output in list format 
    print(content)
```

```python
#Read and write mode
with open("readwrite.txt",mode="r+") as f:
    #Writing in the file
    content=f.write("Learning file handling read and write\n")
```

```python
#Read and write mode
with open("readwrite.txt",mode="r+") as f:
    #Reading the file
    content=f.read()
    print(content)

    #Writing in the file
    content=f.write("Learning file handling writing now\n")
```

```python
# Write and Read Mode
with open("writeread.txt",mode="w+") as f:
    #Writing the file
    print(f.tell())
    content=f.write("Writing first\n")
    print(f.tell())
    f.seek(0)
    print(f.tell())

    #Reading the file
    content=f.read()
    print(content)
```

```python
# Append and Read Mode
with open("writeread.txt",mode="a+") as f:
    print(f.tell())
    #Append the file
    content=f.write("Appending now\n")
    print(f.tell())
    f.seek(0)
    
    #Reading the file
    content=f.read()
    print(content)
```

```python
#CSV File Handling

content=[["Sonu",23],["Monu",23],["Tonu",34]]
#write in csv file
import csv

#1.opening file
with open("student.csv",mode="w",) as f:
    data=csv.writer(f)
    data.writerow(["Name","Age"]) #single row
    data.writerows(content) #multiple row
```

```python
#CSV File Handling

#1.opening file
with open("student.csv",mode="a",) as f:
    data=csv.writer(f)
    data.writerow(["Name","Age"]) #single row
    # data.writerows(["Vonu",12]) #multiple row
    data.writerows(["Vonu",12]) #multiple row
```

```python
#CSV File Handling
#Reading CSV
with open("student.csv",mode="r",) as f:
    data=csv.reader(f)
    for row in data:
        print(row)
```
