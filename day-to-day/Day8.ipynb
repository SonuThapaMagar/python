```python
#Sync and asyunc function
import time

def petrol_pump(v_id):
    print(f"Turn of {v_id}")
    time.sleep(2)
    print(f"Filling petrol: {v_id}")
    time.sleep(5)
    print(f"Paid cash: {v_id}")

# petrol_pump(123)
```

```python
def main():
    start_time=time.time() # start time initialize 0 sec
    petrol_pump(123)
    petrol_pump(234)
    end_time=time.time() # 
    print(f"Total execution: {end_time -start_time:.2f}sec")
```

```python
main()
```

```python
import asyncio
```

```python
#Asynchronous function
async def petrol_pump(v_id):
    print(f"Turn of {v_id}")
    await asyncio.sleep(2)
    print(f"Filling petrol {v_id}")
    await asyncio.sleep(5)
    print(f"Paid cash:{v_id}")
```

```python
async def main():
    start_time=time.time()
    await asyncio.gather(petrol_pump(123),petrol_pump(234))
    end_time=time.time()
    print(f"Total execution :{end_time-start_time:.2f}sec")
```

```python
await main()
```

```python
#Decorators
def outer(main):  # function as param
    def inner(): 
        main() # function inside function
        print("World")
    return inner # return function
```

```python
#Decorators
@outer
def main():
    print("Hello")
```

```python
main()
```

```python
def decorator_func(func):
    def inner(a,b):
        a+=5
        sum=func(a,b)
        return sum

    return inner
```

```python
@decorator_func
def cal_sum(a,b):
    return a+b
```

```python
cal_sum(2,3)
```
