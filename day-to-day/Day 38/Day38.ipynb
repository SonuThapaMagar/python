```python
def df(x):
    return 4*x+3
```

```python
def f(x):
    return 2*x**2+3*x+1
```

```python
def gradient_descent(df,x,lr):
    iter=0;
    max_iter=100
    while iter<=max_iter:
        x=x-lr*df(x)
        iter+=1
        optimal_x=x
        return optimal_x,iter
    
opt_x,strp=gradient_descent(df,0.01,lr)
opt_x
```
