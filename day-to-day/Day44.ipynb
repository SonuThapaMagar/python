```python

x=torch.tensor([[1.0,2.0,3.0]])
y=torch.tensor([[2.5]])
```

```python
class NeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        
        self.input=nn.Linear(3,4) #3 input and 4 output
        self.hidden=nn.Linear(4,3) # 4 input, 3 output
        self.output=nn.Linear(3,1) #3 input, 1 output
        
    def forward(self,x):
        z1=self.input(x)
        alpha1=torch.relu(z1)
        
        z2=self.hidden(alpha1)
        alpha2=torch.relu(z2)
        
        z3=self.output(alpha2)
        
        y_pred=torch.relu(z3)
        return y_pred
```

```python
model=NeuralNetwork()
print(model(x))
```

```python
los=nn.MSELoss()
optimizer=torch.optim.Adam(model.parameters(),lr=0.01,betas=(0.9,0.999))
```

```python
loss_history=[]
```

```python
for i in range(1,500):
    y_pred=model(x)
    loss=los(y_pred,y)
    loss_history.append(loss.item())
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    print(f"Epoch {i},Prediction={y_pred.item()}, Loss={loss.item()}")
print(f"\nFinal Prediction: ", model(x).item())
print("Target",y.item())
```
