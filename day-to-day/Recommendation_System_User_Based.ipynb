```python
sparsity=1-density
print("Sparsity: {:.2%}".format(sparsity))
```

Average Rating

```python
user_mean=user_item.mean(axis=1)
user_mean.head()
```

Mean Centering

```python
user_item_centered=user_item.sub(user_mean,axis=0).fillna(0) #not rated movie ko lagi fillna(0) 0 
user_item_centered.iloc[:5,:5]
```

```python
centered_values=user_item_centered.values
centered_values
```

Rated movies- 1, non rated - 0

```python
rated_mask=user_item.notna().values.astype(float)
rated_mask
```

```python
movie_ids=user_item.columns
print(movie_ids)
user_ids=user_item.index
print(user_ids)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

#user user similarity calculation
user_similarity=cosine_similarity(user_item_centered)
np.fill_diagonal(user_similarity,0)
print("Similarity Shape: ",user_similarity.shape)
```

```python
user_sim_df=pd.DataFrame(user_similarity,index=user_ids,columns=user_ids)
user_sim_df.iloc[:5,:5]
```

```python
#most similar user to user 1
target_user=1
user_sim_df[target_user].sort_values(ascending=False).head()
```

```python
similar_user=user_sim_df[target_user].idxmax()
print("Most similar user:", similar_user)
```

```python
common=movie_ratings[movie_ratings['userId'].isin([target_user,similar_user])]
common=common.pivot_table(index='title',columns='userId',values='rating').dropna()
common.head(10)
```

```python
u=user_ids.get_loc(1)#user id=1
sims=user_similarity[u] #location or index of user 1 is 0
print(u)
print(sims)
```

```python
top=np.argsort (sims)[::-1][:20]
top #index of movie id
```

```python
print(centered_values[top])
```

```python
w=sims[top]
w
```

```python
num= w @ centered_values[top]
num
```

```python
den=np.abs(w) @ rated_mask[top]
den
```

```python
support=rated_mask[top].sum(axis=0)
print(support)
```

```python
print("Num",num) # num= w @ centered_values[top]

print("Density:", den) #den=np.abs(w) @ rated_mask[top]
```

```python
pred=user_mean.iloc[u] + np.divide(num,den,out=np.zeros_like(num), where=den>0)
pred[support<3]=np.nan # user ko min 3 neighbors 
print(pred)
print(pred.shape)
```

```python
def predict_for_user(user_id,k=50,min_neighbors=3):
    u=user_ids.get_loc(user_id)
    sims=user_similarity[u]
    top=np.argsort(sims)[::-1][:k]
    top=top[sims[top]>0]
    w=sims[top]
    num= w @ centered_values[top]
    den=np.abs(w) @ rated_mask[top]
    support=rated_mask[top].sum(axis=0)

    pred=user_mean.iloc[u] + np.divide(num,den,out=np.zeros_like(num), where=den>0)
    pred[support<min_neighbors]=np.nan # user ko min 3 neighbors 
    
    return pd.Series(pred,index=movie_ids)
```

```python
#Predicted ratings for user 1
predict_for_user(1).dropna().sort_values(ascending=False).head()
```
