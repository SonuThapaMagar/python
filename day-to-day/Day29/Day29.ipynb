```python
from scipy.stats import skew,gaussian_kde
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
file_path="real_estate_valuation.csv"

#To read csv file
df=pd.read_csv(file_path)
```

```python
df.columns
```

```python
cols=df.columns.tolist()
cols
```

```python
print(skew(df))
```

```python
for col in cols:
    plt.figure(figsize=(5,6))
    sns.displot(df[col],kde=True)
    plt.show()
```

```python
import numpy as np
print("Log Transformation")
      
for col in cols:
    log_data=np.log(df[col])
    print(f" After transformation {df}")
    print(skew(log_data))
    plt.figure(figsize=(6,4))
    sns.displot(df[col],kde=True)
    plt.show()
```

```python
import numpy as np
print("Log P Transformation")
      
for col in cols:
    log_p_data=np.log1p(df[col])
    print(f" After log p transformation {df}")
    print(skew(log_p_data))
    plt.figure(figsize=(6,4))
    sns.displot(df[col],kde=True)
    plt.show()
```

```python
import numpy as np
print("Log Transformation")
      
for col in cols:
    log_data=np.sqrt(df[col])
    print(f" After transformation {df}")
    print(skew(log_data))
    plt.figure(figsize=(6,4))
    sns.displot(df[col],kde=True)
    plt.show()
```

```python
from scipy.stats import boxcox,yeojohnson

print("Yeo Johnson Transformation")
      
for col in cols:
    data,l=yeojohnson(df[col])
    print(f" After transformation {df}")
    # print(skew(data))
    
    plt.figure(figsize=(6,4))
    sns.displot(df[col],kde=True)
    plt.show()
```
