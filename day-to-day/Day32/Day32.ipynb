```python
from sklearn.preprocessing import PowerTransformer
from sklearn.compose import ColumnTransformer

trf_yeo=ColumnTransformer([("transform_numerical",PowerTransformer(method="yeo-johnson"),numerical_columns)],remainder="passthrough",verbose_feature_names_out=False).set_output(transform="pandas")
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")

trf_imputer = ColumnTransformer(
    [
        (
            "imputation",
            SimpleImputer(strategy="median"),
            numerical_columns
        )
    ],
    remainder="passthrough",
    verbose_feature_names_out=False
).set_output(transform="pandas")
```

```python
df = trf_imputer.fit_transform(df)
df.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(sparse_output=False,dtype = int)
encoded = encoder.fit_transform(df[["ocean_proximity"]])
df[encoder.get_feature_names_out()] = encoded
df.head()
```

```python
df.drop(columns = ["ocean_proximity"],inplace=  True)
df
```

```python
from sklearn.preprocessing import MinMaxScaler

trf_scaler = ColumnTransformer(
    [
        (
            "scaling",
            MinMaxScaler(),
            numerical_columns
        )
    ],
    remainder="passthrough",
    verbose_feature_names_out=False
).set_output(transform="pandas")

df = trf_scaler.fit_transform(df)
df.head()
```

```python
from sklearn.pipeline import Pipeline

pipe = Pipeline([
    ("imputation", trf_imputer),
    ("yeo_johnson", trf_yeo),
    ("scaling", trf_scaler)
])
```
