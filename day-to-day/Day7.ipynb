```python
#Function
def calc_sum(a,b,c=6):
    sum=a+b+c
    return sum

s=calc_sum(2,4,4)
print(s)
```

```python
#*args
def add_all(*nums):
    return sum(nums)

add_all(5,5,2)
```

```python
#*args
def user_info(*infos):
    print(type(infos))
    for info in infos:
        print(f"User Info: {info}")

user=user_info(("Sonu",10),("Kritika",23),("Chesta",34))
```

```python
#*args
def user_info(*names):
    print(names)
    for name,id in names:
            print(f"Name: {name}, Id:{id}")

user=user_info(("Sonu",10),("Kritika",23),("Chesta",34))
```

```python
#**kwargs
def user_info(**infos):
    print(f"User name is {infos["name"]}, age is {infos["age"]} and her department is {infos["faculty"]}")
    for key,value in infos.items():
        print(f"{key}:{value}")
print(user_info(name="Sonu",age=23, faculty="IT")) 
```

```python
#Global Variable
x=5

def scope():
    # local
    a=34
    print(x)
    print(a)

scope()
print(x)
print(a)
```

```python
#Global Variable declaration inside function
x=5

def scope():
    global a
    a=34
    print(f"Inside function: {x}")
    print(f"Inside function: {a}")

scope()
print(f"Outside function: {x}")
print(f"Outside function: {a}")
```
