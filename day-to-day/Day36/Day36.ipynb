```python
x = df[["TV","radio"]].values
X = np.c_[np.ones(df.shape[0]),x]
X
```

```python
Y = df[["sales"]].values
```

```python
def mlr(x,y):
  coff_b =np.linalg.inv(x.T @ x) @ x.T @ y
  return coff_b
```

```python
b = mlr(X,Y)
b.shape
```

```python
y_pred = X @ b
y_pred
```

```python
class MultipleLinearRegression:
  def __init__(self):
    self.coff_ = None
  def fit(self,x,y):
    X = np.c_[np.ones(x.shape[0]),x]
    self.coff_ = np.linalg.inv(X.T @ X) @ X.T @ y
    return self
  def predict (self,x):
    X = np.c_[np.ones(x.shape[0]),x]
    return X @ self.coff_

```

```python
xa = df[["TV","radio"]].values
ya = df[["sales"]].values
```

```python
model = MultipleLinearRegression()
model.fit(xa,ya)
```

```python
y_pred = model.predict(xa)
y_pred
```

```python
import numpy as np
x = np.linspace(1,10,8)
```

```python
y = np.sin(x) + np.random.random(8)
```

```python
plt.scatter(x,y)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
m1 = LinearRegression()
x = x.reshape(-1,1)
y = y.reshape(-1,1)

m1.fit(x,y)
y_pred = m1.predict(x)
```

```python
plt.scatter(x,y)
plt.plot(x,y_pred,color = "red")
plt.show()
```

```python
from sklearn.preprocessing import PolynomialFeatures
for d in range(1,10):
  poly  = PolynomialFeatures(degree=d)
  poly_x = poly.fit_transform(x)
  m1.fit(poly_x,y)
  y1 = m1.predict(poly_x)
  plt.scatter(x,y)
  plt.plot(x,y1,color = "red")
  plt.show()
```
