```python
import re

text="apple appplee appl"

print(re.findall(r"e",text))
```

```python
text="sonumagar787@gmail.com, abc#mail.com, abc@gmail.com"

print(re.findall(r"@",text))
print(re.sub(r"\s","",text))
```

```python
text="""
cat, mat, caet,
c9t,caat,cat
"""

print(re.findall(r"c.t",text))
```

```python
text="""
HELLO!!! Visit https://example.com
Contact: student123@gmail.com
I LOVE NEPAL????>
I am learning Python @student.
"""
```

```python
text=text.lower()
print(text)
```

```python
text=re.sub(r"https?://\S+|www\.\S+","",text)
print(text)
```

```python
text=re.sub(r"\b[\w.+--]+@[\w.-]+\.[A-Za-z]{2,}\b","",text)
print(text)
```

```python
text=re.sub(r"[^a-zA-Z0-9\s]","",text)
print(text)
```
