```python
#Polymorphism method overloading
class Student:
    def __init__(self,name,roll):
        self.name=name
        self.roll=roll

    def __add__(self, other): #self = s1 obj value , other= s2 obj value
        name3=self.name+other.name
        roll3=self.roll+other.roll
        s3=Student(name3,roll3) # s3 object only forms after calling Class:Student where new values are passed for addition
        return s3

    def __str__(self):
        return (
            f"Name: {self.name}\n"
            f"Roll : {self.roll}"
        )
```

```python
s1=Student("Sonu",11)
s2=Student("Monu",22)
s3=s1+s2
print(s3)
```

```python
#Polymorphism method overloading
class Student:
    def __init__(self,mark1,mark2):
        self.mark1=mark1
        self.mark2=mark2

    def __add__(self, other): #self = s1 obj value , other= s2 obj value
        result1=self.mark1 & other.mark1
        result2=self.mark2 & other.mark2
        s3=Student(result1,result2) # s3 object only forms after calling Class:Student where new values are passed for addition
        return s3

    def __str__(self):
            return (
                f"{self.mark1} {self.mark2}"
            )
```

```python
s1=Student(4,3)
s2=Student(3,4)
```

```python
s3=s1 + s2
print(s3)
```

```python
#Polymorphism method overloading greater than
class CheckValue:
    def __init__(self,num):
        self.num=num

    def __gt__(self, other): #self = s1 obj value , other= s2 obj value
        if(self.num > other.num):
             return True
        else:
             return False
```

```python
n1=CheckValue(9)
n2=CheckValue(3)
```

```python
n3=n1 > n2
print(n3)
```

```python
#Object as a function __call__=magic function
class Mul:
    def __init__(self,num):
        self.num=num

    def __call__(self, num2): 
        return self.num * num2 # self.num=3, num2 =2
```

```python
n1=Mul(3)
n1.num
n1(2) # object call as a function
```

```python
class Person:
    def __init__(self, name):
        self.name = name

    def __add__(self, other):
        return Person(self.name + other.name)


father = Person("Ram")
mother = Person("Sita")

child = father + mother

print(child.name)
```
