```python
import pandas as pd
import numpy as np

file_path = "data_scaling_skewed_data.csv"
df = pd.read_csv(file_path)

df.head()
```

```python
df.shape
```

```python
def osl(x,y):# x independent ,y - dependent
    x_mean=np.mean(x)
    y_mean=np.mean(y)
    Nr=0
    Dr=0
    
    n=len(x)
    for i in range(n):
        Nr+=(x[i]-x_mean)*(y[i]-y_mean)
        Dr+=(x[i]-x_mean)**2
        
    m=Nr/Dr
    b=y_mean-(m*x_mean)
    y_predict=m*x+b
    print(f"Slope m: {m}")
    print(f"Y intercept b: {b}")
    return m,b,y_predict
```

```python
x=df["Study_Hours"].values
x
```

```python
y=df["Exam_Score"].values
y
```

```python
osl(x,y)
```
