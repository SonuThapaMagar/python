```python
import pandas as pd
import numpy as np

file_path = "data_scaling_skewed_data.csv"
df = pd.read_csv(file_path)

df.head()
```

```python
df.shape
```

```python
def osl(x,y):# x independent ,y - dependent
    x_mean=np.mean(x)
    y_mean=np.mean(y)
    Nr=0
    Dr=0
    
    n=len(x)
    for i in range(n):
        Nr+=(x[i]-x_mean)*(y[i]-y_mean)
        Dr+=(x[i]-x_mean)**2
        
    m=Nr/Dr
    b=y_mean-(m*x_mean)
    y_predict=m*x+b
    print(f"Slope m: {m}")
    print(f"Y intercept b: {b}")
    return m,b,y_predict
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.scatter(df["Study_Hours"],df["Exam_Score"],color="Red",label='Data Points')
plt.title("Scatter Plot of Data Study Hour and Exam Score")

plt.xlabel("Data Study Hour")
plt.ylabel("Data Exam Score")

plt.grid(True)
plt.legend()

plt.show()
```

```python
x=df["Study_Hours"].values
x
```

```python
y=df["Exam_Score"].values
y
```

```python
m,b,y_pred=osl(x,y)
```

```python
y_pred
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.scatter(df["Study_Hours"],df["Exam_Score"],color="Red",label='Data Points')
plt.plot(df["Study_Hours"],y_pred,color="black")
plt.title("Scatter Plot of Data Study Hour and Exam Score")

plt.xlabel("Data Study Hour")
plt.ylabel("Data Exam Score")

plt.grid(True)
plt.legend()

plt.show()
```

```python
x=df[["Study_Hours"]].values
y=df["Exam_Score"].values
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split # model test and train

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
#for prediction
model.coef_[0] #m
```

```python
model.intercept_ #y
```

```python
#for prediction
y_pred=model.predict(x_test)
y_pred
```

```python
compare=pd.DataFrame({
    "Actual": y_test,
    "Predict":y_pred
})
```
