```python
num[0:5]
```

```python
#2D Array
num=np.arange(24).reshape(4,6)
num
```

```python
num[0,3:]
```

```python
num[0:5:2,1:4:2]
```

```python
#3D Array
a=np.arange(60).reshape(3,4,-1) #also works (3,4,5) 
a
```

```python
a[1][2]
```

```python
#21,23,31,33
a[1,0:3:2,1:4:2]
```

```python
a[[1,2],[3,0],:]
```

```python
a.reshape(-1) #reshape helps in flattening
```

```python
a=np.arange(12).reshape(3,4)
np.mean(a)
print(a)
```

```python
np.mean(a,axis=0) #axis=0 meaning col
```

```python
np.mean(a,axis=1) #axis=0 meaning row
```

```python
np.median(a)
np.median(a,axis=0) 
```

```python
np.median(a)
np.median(a,axis=1) 
```

```python
np.var(a)
```

```python
np.std(a,axis=0)
```

```python
np.std(a,axis=1)
```
