```python
from scipy.stats import skew,gaussian_kde
import numpy as np

a=[2,3,3,4,5,6,7,8,25]
b=[1,15,16,17,18,19,20,21,22]
c=[10,12,15,15,15,16,18,20]


s_a=skew(a)
s_b=skew(b)
s_c=skew(c)


print(s_a)
print(s_b)
print(s_c)

g_a=gaussian_kde(a)
g_b=gaussian_kde(b)
g_c=gaussian_kde(c)

plt.figure(figsize=(9,6))
xa_range=np.linspace(min(a)-2,max(a)+2,100)
ya_range=g_a(xa_range)
plt.plot(xa_range,ya_range,color="Red")

xb_range=np.linspace(min(b)-2,max(b)+2,100)
yb_range=g_b(xb_range)
plt.plot(xb_range,yb_range,color="Green")

xc_range=np.linspace(min(c)-2,max(c)+2,100)
yc_range=g_c(xc_range)
plt.plot(xc_range,yc_range,color="Blue")
plt.show()
```

```python
x=[1,2,3,4,5]
y=[2,4,6,8,10]

import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.show()
plt.plot(x,y,color="green",marker="o",markersize=4,linewidth="2",linestyle=":",label="Linear Graph")
plt.legend() #must be after the label definition
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Linear Graph")
plt.show()
```

```python
x=[1,2,3,4,5,6,7]
y=[2,5,9,8,10,2,3]

import matplotlib.pyplot as plt

plt.figure(figsize=(9,5))
plt.show()
plt.scatter(x,y,color="green")
plt.legend() #must be after the label definition
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Linear Graph")
plt.show()
```

```python
#Bar Graph
x=[1,2,3,4,5,6,7]
y=[2,5,9,8,10,2,3]

import matplotlib.pyplot as plt

plt.figure(figsize=(9,5))
plt.show()
plt.bar(x,y,color="green")
plt.legend() #must be after the label definition
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Linear Graph")
plt.show()
```
