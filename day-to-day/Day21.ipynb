```python
from scipy.stats import kurtosis,gaussian_kde
import matplotlib.pyplot as plt
import numpy as np

a=[4.2,5.1,4.8,5.5,4.9,5.0,5.3,4.7,5.2,4.6,5.4,4.5,5.1,4.9,5.0,5.2,4.8,5.3,4.7,5.1]

b=[5.0,5.1,4.9,5.0,5.05,4.95,5.0,5.02,4.98,5.0,5.01,4.99,5.0,2.0,8.0,1.5,8.5,5.0,5.0,5.0]

c=[3.0,3.5,4.0,4.5,5.0,5.5,6.0,6.5,7.0,3.2,3.8,4.2,4.8,5.2,5.8,6.2,6.8,7.2]


k_a=kurtosis(a)
k_b=kurtosis(b)
k_c=kurtosis(c)

g_a=gaussian_kde(a)
g_b=gaussian_kde(b)
g_c=gaussian_kde(c)

plt.figure(figsize=(9,6))
plt.plot(label='Kurtosis Graph')
xa_range=np.linspace(min(a)-2,max(a)+2,100)
ya_range=g_a(xa_range)
plt.plot(xa_range,ya_range,color="Red")

xb_range=np.linspace(min(b)-2,max(b)+2,100)
yb_range=g_b(xb_range)
plt.plot(xb_range,yb_range,color="Green")

xc_range=np.linspace(min(c)-2,max(c)+2,100)
yc_range=g_c(xc_range)
plt.plot(xc_range,yc_range,color="Blue")
plt.show()
```

```python
#Q2: Find z score of 71 and 42 from data 
import numpy as np

a=[25,36,42,53,63,71,82,88,89,90]

#Mean
mean_a=np.mean(a)
print(f"Mean: {mean_a}")

#Standard Deviation
std_a=np.std(a)
print(f"Standard Deviation: {std_a:.2f}")

#Z score fo 71
z_score_71=(71-mean_a)/std_a
print(f"Z score of 71: {z_score_71:.2f}")

z_score_42=(42-mean_a)/std_a
print(f"Z score of 42: {z_score_42:.2f}")
```

```python
#Q3; Find the outlier and plt graph of [1,12,13,14,15,17,25] without using seaborn boxplot
# from scipy.stats import 

a=np.array([1,12,13,14,15,17,25])

q1=np.percentile(a,25)
q3=np.percentile(a,75)

print(f"Q1:{q1}\n Q3:{q3}")

iqr=q3-q1
print(f"IQR: {iqr}")

lf=q1-1.5*iqr
uf=q3+1.5*iqr
print(f"Lower fence: {lf} \n Upper fence:{uf}")

#Outlier
outlier_mask=(a<lf)| (a>uf)
print(f"Outlier: {outlier_mask}")


outlier_data=a[outlier_mask]
print(f"Outlier data are:{outlier_data}")

clean_data=a[~outlier_mask]
print(f"Clean data is {clean_data}")

x=np.arange(len(a))
```

```python
plt.figure(figsize=(10,6))
plt.scatter(x[outlier_mask],outlier_data,color="Red",label='Outlier Data')
plt.scatter(x[~outlier_mask],clean_data,color="blue",label='Clean Data')

plt.axhline(lf,color="green",ls=":",label="Lower Fence")
plt.axhline(uf,color="black",ls=":",label="Upper Fence")

plt.axhline(q1,color="magenta",ls="--",label="Quartile 1")
plt.axhline(q3,color="magenta",ls="--",label="Quartile 3")

plt.legend()
plt.grid(True)
plt.show()
```

```python
#Covariance and correlation
a = np.array([2, 4, 5, 6, 7, 8, 9, 10])
b = np.array([55, 60, 65, 70, 75, 80, 85, 90] )

# Covariation
covariation_first=np.cov(a,b)
print(f"Covariance: {covariation_first}")

#Correlation
correlation_first=np.corrcoef(a,b)
print(f"Correlation:{correlation_first}")

plt.figure(figsize=(10,6))
plt.scatter(a,b,color="Red",label='Data')
plt.show()
```
