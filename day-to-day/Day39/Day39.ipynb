```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import classification_report, accuracy_score

# Base Models
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
```

```python
# 1. LOAD DATA
iris=load_iris()
```

```python
X = iris.data  
y = iris.target
```

```python
# 2. SPLIT FIRST (Crucial for preventing data leakage)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=10, shuffle=True
)
```

```python
# 3. PREPROCESS (Scale based ONLY on training data)
scaler=MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test) 
```

```python
# 4. MODEL DICTIONARY
models = {
    "Logistic Regression": LogisticRegression(max_iter=200),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Naive Bayes": GaussianNB(),
    "SVM": SVC(probability=True, random_state=42)
}
```

```python
# 5. TRAINING AND EVALUATION LOOP
for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)
    print(f"{name:<20}: {accuracy_score(y_test, y_pred):.4f}")
```

```python
# 6. CORRECTLY PREDICTING A SINGLE NEW DATA POINT
single_test_point = np.array([[0.67777, 0.21333, 0.5433, 0.822]])
single_test_scaled = scaler.transform(single_test_point)
```

```python
knn_model = models["KNN"]
```

```python
prediction = knn_model.predict(single_test_scaled)
predicted_class = iris.target_names[prediction[0]]
probabilities = knn_model.predict_proba(single_test_scaled)
```

```python
print(f"Predicted Class: {predicted_class}")
print(f"Probabilities (Setosa, Versicolor, Virginica): \n{probabilities}")
```
