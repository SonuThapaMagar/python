```python
from scipy.stats import bernoulli,binom
import numpy as np
```

```python
from scipy.stats import bernoulli

p=0.7

success=bernoulli.pmf(k=1,p=0.7)
failure=bernoulli.pmf(k=0,p=0.7)
print(success)
print(failure)
```

```python
from scipy.stats import bernoulli

p=0.7

p1=bernoulli.cdf(k=1,p=0.7)
p2=bernoulli.cdf(k=0,p=0.7)
p3=bernoulli.cdf(k=-1.2,p=0.7)
print(p1)
print(p2)
print(p3)
```

```python
n=10
k=5
p=0.5
prob=binom.pmf(k,n,p)
print(prob)
```

```python
n=10
k=np.arange(0,n+1)
p=0.5
prob=binom.cdf(k,n,p)
print(prob)
```

```python
n=10
k=np.arange(0,n+1)
p=0.1
prob=binom.pmf(k,n,p)
print(prob)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.bar(k,prob,color="Blue") #k= x axis , prob= y axis
plt.show()
```

```python
#cdf 
k=np.arange(0,5)

prob=binom.cdf(k,n,p)
print(prob)
```
