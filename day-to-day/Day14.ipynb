```python
#Composition
class Customer:
    def __init__(self,name,location):
        self.name=name
        self.location=location

    def c_info(self):
        return (
            f"Customer Name: {self.name}\n from"
            f"Location: {self.location} has ordered"
        )
    
class Product:
    def __init__(self,item,price):
        self.item=item
        self.price=price

    def p_info(self):
        return (
            f"Product Name: {self.item}\n costing"
            f"Price: {self.price}"
        )

class Order:
    def __init__(self,name,location,item,price,order_id):
        self.obj_customer=Customer(name,location)
        self.obj_product=Product(item,price)
        self.order=order_id

    def o_info(self):
        print(self.obj_customer.c_info())
        print(self.obj_product.p_info())
        return (
            f"Order id : {self.order}"
        )
```

```python
order=Order("Sonu","Bhaisepati","Laptop",2500,2345)
order.o_info()
```

```python
#Aggregation
class Employee:
    def __init__(self,name,emp_id):
        self.name=name
        self.emp_id=emp_id

    def e_info(self):
        return (
            f"Name: {self.name} having\n"
            f"Employee Id: {self.emp_id} is from"
        )

class Department:
    def __init__(self,dname,obj_e):
        self.dname=dname
        self.obj_e=obj_e

    def d_info(self):
        print(self.obj_e.e_info())
        return f"Department: {self.dname}"
```

```python
e1=Employee("Sonu",123)
d1=Department("IT",e1)
```

```python
d1.d_info()
```

```python
#Abstract Class Method
from abc import ABC,abstractmethod
class Payment(ABC): #Abstract Base Class

    @abstractmethod #Decorator
    def payment(self):
        pass #abstract ma no logic implementation

class Esewa(Payment):
    def payment(self):
        # return super().payment()
        return f"Payment made through esewa"

class Khalti(Payment):
    def payment(self):
        # return super().payment()
        return f"Payment made through Khalti"
```

```python
e1=Esewa()
e1.payment()
```

```python
k1=Khalti()
k1.payment()
```
