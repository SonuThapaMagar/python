Stop-word removal

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download the required word lists and tokenizer data
nltk.download("stopwords")
nltk.download("word_tokenize")
```

```python
text="The cat is sitting on the mat."

words=word_tokenize(text)
words
```

```python
stop_words = set(stopwords.words("english"))
stop_words
```

```python
filtered_words = [
    word for word in words
    if word.lower() not in stop_words
]

print(filtered_words)
```

Stemming and Lemmatization

```python
# Download dictionary data for lemmatization
nltk.download("wordnet")
```

```python
from nltk.stem import PorterStemmer,WordNetLemmatizer

#Creating tools
stemmer =PorterStemmer()
lemmatizer=WordNetLemmatizer()
```

```python
text = "The software engineer studies multiple programming languages. She is building data pipelines and creating machine learning models!"

#Tokenize
words=word_tokenize(text)

# Apply stemming
stemmed_words=[stemmer.stem(w) for w in words]

# Apply lemmatization
lemmatized_words=[lemmatizer.lemmatize(w, pos="v") for w in words]
```

```python
print(words)
print(stemmed_words)
print(lemmatized_words)
```

Bag of Words (BoW)

```python
from sklearn.feature_extraction.text import CountVectorizer

documents=[
    "I love NLP",
    "I love AI",
    "NLP is fun"
]

#Create BoW vectorizer
vectorizer=CountVectorizer() 
#CountVectorizer ignores single-character tokens such as I. It also sorts the feature names alphabetically.

#Convert doc in numerical word counts
X=vectorizer.fit_transform(documents)

#Displacy vocab
print("Vocabulary:")
print(vectorizer.get_feature_names_out())

#Matrix
print("\nDocument-Term Matrix:")
print(X.toarray())
```

Bag of Words(BoW) using n-gram


```python
#Using n-gram
vectorizer=CountVectorizer(ngram_range=(1,2))

#Convert doc in numerical word counts
X=vectorizer.fit_transform(documents)

#Displacy vocab
print("Vocabulary:")
print(vectorizer.get_feature_names_out())

#Matrix
print("\nDocument-Term Matrix:")
print(X.toarray())
```

TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "cat cat dog",
    "dog and mouse",
    "cat and mouse"
]

#Create TF-IDF vectorizer
vectorizer=TfidfVectorizer()

#Learn the vocab abd calculate TF-IDF scores
X=vectorizer.fit_transform(documents)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(X.toarray().round(3))
```
