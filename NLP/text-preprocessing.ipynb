Text normalization

```python
#Original test
text="  Hello ! My name is Sonu Thapa Magar. I am learning AI/Machine Learning at WIA Nepal !."
```

```python
#Converting to lower case
text=text.lower()
text
```

```python
#Removing spaces from beginning
text=text.strip()
text
```

Sentence Segmentation

```python
#Splitting paragraph at each period
sentences=text.split(".")
sentences
```

```python
# Remove extra spaces and empty items
sentences=[sentence.strip() for sentence in sentences if sentence.strip()]
sentences
```

Tokenization

```python
#Splitting text into words
tokens=text.split()
tokens
```

```python
#Length
print(len(tokens))
```

Punctuation Handling

```python
import string

sentences ="  Hello ! My name is Sonu Thapa Magar. I am learning AI/Machine Learning at WIA Nepal !."
for symbol in string.punctuation:
    sentences=sentences.replace(symbol,"");
    
sentences
```

Stop-word removal

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download the required word lists and tokenizer data
nltk.download("stopwords")
nltk.download("word_tokenize")
```

```python
text="The cat is sitting on the mat."

words=word_tokenize(text)
words
```

```python
stop_words = set(stopwords.words("english"))
stop_words
```

```python
filtered_words = [
    word for word in words
    if word.lower() not in stop_words
]

print(filtered_words)
```
