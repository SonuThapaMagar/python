```python
skewed_columns = [
    "MSSubClass",
    "LotFrontage",
    "LotArea",
    "MasVnrArea",
    "BsmtFinSF1",
    "BsmtFinSF2",
    "TotalBsmtSF",
    "1stFlrSF",
    "LowQualFinSF",
    "GrLivArea",
    "BsmtHalfBath",
    "KitchenAbvGr",
    "WoodDeckSF",
    "OpenPorchSF",
    "EnclosedPorch",
    "3SsnPorch",
    "ScreenPorch",
    "PoolArea",
    "MiscVal",
    "SalePrice",
]
```

```python
from sklearn.preprocessing import PowerTransformer


pt=PowerTransformer(method="yeo-johnson")

df[skewed_columns]=pt.fit_transform(df[skewed_columns])
```
