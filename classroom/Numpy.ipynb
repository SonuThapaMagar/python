```python
import numpy as np
```

```python
# 1D array
a = np.array([1,2,3,4,5,6,7,8])
print(a)
```

```python
a[0:5]
```

```python
a[-4:-1]
```

```python
#2D array
a = np.arange(24).reshape(4,6)
a
```

```python
a[0:5:2,1:4:2]
```

```python
# 3D Array
a =  np.arange(60).reshape(3,4,-1)
a
```

```python
a[[0,1],[3,0],:]
```

```python
a = np.arange(12).reshape(3,4)
a
```

```python
np.mean(a,axis = 1) # axis  =0 col,1 row
```

```python
np.median(a,axis=  1)
```

```python
np.var(a,axis =1)
```

```python
np.std(a)
```
