```python
x = df[["TV","radio"]].values
X = np.c_[np.ones(df.shape[0]),x]
X
```

```python
Y = df[["sales"]].values
```

```python
def mlr(x,y):
  coff_b =np.linalg.inv(x.T @ x) @ x.T @ y
  return coff_b
```

```python
b = mlr(X,Y)
b.shape
```

```python
y_pred = X @ b
y_pred
```
