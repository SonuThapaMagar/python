```python
with open("abc.txt",mode = "w",buffering = -1) as f:
    content = ["Hi\n","Hello\n","How are you\n"]
    f.writelines(content)
    f.write("I am Learning Python\n")
    
            
```

```python
with open("abc.txt",mode = "r") as f:
    # content = f.read()
    # content = f.readline()
    content = f.readlines()
    print(content)
    f.write("Hi")
```

```python
with open("abc.txt",mode = "r+") as f:
    content = f.read()
    print(content)
    f.write("Hi2\n")
    
```

```python
with open("abc.txt",mode = "w+") as f:
    print(f.tell())
    f.write("Hello world")
    print(f.tell())
    f.seek(0)
    print(f.tell())
    content = f.read()
    print(content)
    print(f.tell())
```

```python
with open("abc.txt",mode=  "a+") as f:
    print(f.tell())
    f.write("There")
    print(f.tell())
    f.seek(0)
    print(f.read())
    
```

```python
# csv file
# writee in csv file
content = [["Rama",12],["Saman",13]]
import csv
with open("student.csv",mode=  "w") as f:
    data = csv.writer(f)
    data.writerow(["Name","Rollno"])
    data.writerows(content)
    
```

```python

with open("student.csv",mode=  "a") as f:
    data = csv.writer(f)
    data.writerow(["Deep",14])
```

```python
with open("student.csv", mode=  "r") as f:
    data = csv.reader(f)
    for row in data:
        print(row)
```
