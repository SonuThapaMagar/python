```python
class MultipleLinearRegression:
  def __init__(self):
    self.coff = None

  def fit(self,x,y):
    x = np.c_[np.ones(x.shape[0]),x]
    self.coff =np.linalg.inv(x.T @ x) @ x.T @ y
    return self

  def predict(self,x):
    x_b = np.c_[np.ones(x.shape[0]),x]
    return  x_b @ self.coff


```

```python
from sklearn.linear_model import LassoCV, RidgeCV
from sklearn.preprocessing import PolynomialFeatures
for d in range(1,9):
  poly_feature = PolynomialFeatures(degree=d)
  poly_x = poly_feature.fit_transform(x)

  linear = LinearRegression()
  # l1 = LassoCV(alpha= 10)
  l2 = RidgeCV(alphas = [11,10,22,40],cv=5)

  linear.fit(poly_x,y)
  # l1.fit(poly_x,y)
  l2.fit(poly_x,y)

  y_p  = linear.predict(poly_x)
  # y_l1 = l1.predict(poly_x)
  y_l2 = l2.predict(poly_x)
  print(l2.alpha_)
  plt.scatter(x, y, color='red', label='Data')
  plt.plot(x, y_p, color='blue', label='Linear Prediction')
  # plt.plot(x, y_l1, color='green', label='Lasso Prediction')
  plt.plot(x, y_l2, color='yellow', label='Ridge Prediction')
  plt.title(f"Degree {d}, Alpha {l2.alpha_}")
  plt.xlabel("X")
  plt.ylabel("Y")
  plt.legend()
  plt.show()
```
