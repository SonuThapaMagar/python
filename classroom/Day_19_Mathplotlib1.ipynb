```python
import matplotlib.pyplot as plt
import numpy as np
y = [4,5,3,7,1,3,4]
y_mean = np.mean(y)
plt.figure(figsize=(6,4))
plt.plot(y,color ="Red",marker = "s",markersize =4, linestyle = ":",label = "Linear")
plt.axvline(y_mean,color=  "yellow")
plt.axhline(y_mean,color = "pink")
plt.legend()
plt.xlabel("X value")
plt.ylabel("Y value")
plt.title('Linear Graph')
plt.show()
```
